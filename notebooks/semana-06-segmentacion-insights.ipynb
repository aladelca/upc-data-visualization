```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from _shared import (
    make_base_sales,
    introduce_quality_issues,
    profile_dataframe,
    clean_sales_data,
    build_star_schema,
    save_for_tableau,
    ensure_output_dir,
    contrast_ratio,
    make_high_dimensional_dataset,
)

sns.set_theme(style="whitegrid", context="talk")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
WEEK = "week-06"
OUTPUT_DIR = ensure_output_dir(WEEK)

clean, _ = clean_sales_data(introduce_quality_issues(make_base_sales(n=2200, seed=13), seed=13))
clean['sales_band'] = pd.qcut(clean['sales'], q=4, labels=['Q1 Bajo', 'Q2 Medio-bajo', 'Q3 Medio-alto', 'Q4 Alto'])
clean['high_discount'] = np.where(clean['discount'] >= 0.2, 'High Discount', 'Regular Discount')
clean.head()
```

```python
segment_summary = (
    clean.groupby(['segment', 'high_discount'], as_index=False)
    .agg(total_sales=('sales', 'sum'), avg_profit=('profit', 'mean'), orders=('order_id', 'count'))
)
segment_summary
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
sns.barplot(data=segment_summary, x='segment', y='total_sales', hue='high_discount', palette='Greys', ax=axes[0])
axes[0].set_title('Ventas por segmento y nivel de descuento')
sns.boxplot(data=clean, x='sales_band', y='profit', color='#9ca3af', ax=axes[1])
axes[1].set_title('Profit por banda de ventas')
plt.tight_layout()
```
