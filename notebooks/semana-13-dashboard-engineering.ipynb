```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from _shared import (
    make_base_sales,
    introduce_quality_issues,
    profile_dataframe,
    clean_sales_data,
    build_star_schema,
    save_for_tableau,
    ensure_output_dir,
    contrast_ratio,
    make_high_dimensional_dataset,
)

sns.set_theme(style="whitegrid", context="talk")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
WEEK = "week-13"
OUTPUT_DIR = ensure_output_dir(WEEK)

clean, _ = clean_sales_data(introduce_quality_issues(make_base_sales(n=2600, seed=41), seed=41))
clean['month'] = clean['order_date'].dt.to_period('M').dt.to_timestamp()
```

```python
kpis = pd.DataFrame([
    {'metric': 'Total Sales', 'value': clean['sales'].sum()},
    {'metric': 'Total Profit', 'value': clean['profit'].sum()},
    {'metric': 'Orders', 'value': clean['order_id'].nunique()},
    {'metric': 'Avg Ticket', 'value': clean.groupby('order_id')['sales'].sum().mean()},
])
monthly = clean.groupby('month', as_index=False).agg(monthly_sales=('sales', 'sum'), monthly_profit=('profit', 'sum'))
region_category = clean.groupby(['region', 'category'], as_index=False).agg(total_sales=('sales', 'sum'), avg_profit=('profit', 'mean'))
segment_summary = clean.groupby(['segment', 'channel'], as_index=False).agg(total_sales=('sales', 'sum'), orders=('order_id', 'count'))
[kpis, monthly.head(), region_category.head(), segment_summary.head()]
```

### Actividad principal
Piensa cada tabla como una hoja futura en Tableau:
- `kpis`: portada o resumen ejecutivo
- `monthly`: vista temporal principal
- `region_category`: comparación transversal
- `segment_summary`: filtro o vista contextual


```python
publication_checklist = pd.DataFrame([
    {'check': 'Metricas definidas', 'status': 'Pending'},
    {'check': 'Datos publicables', 'status': 'Pending'},
    {'check': 'Vista principal clara', 'status': 'Pending'},
    {'check': 'Filtros necesarios', 'status': 'Pending'},
    {'check': 'Narrativa de soporte lista', 'status': 'Pending'},
])
publication_checklist
```

```python
for name, df in {
    'kpis': kpis,
    'monthly': monthly,
    'region_category': region_category,
    'segment_summary': segment_summary,
    'publication_checklist': publication_checklist,
}.items():
    save_for_tableau(df, WEEK, name)
```

### Uso teorico de herramientas
- `groupby` modulariza los insumos del dashboard.
- Separar tablas por función mejora claridad y rendimiento en Tableau.
- La checklist obliga a conectar diseño, interactividad y publicación.
