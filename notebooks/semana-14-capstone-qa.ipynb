```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from _shared import (
    make_base_sales,
    introduce_quality_issues,
    profile_dataframe,
    clean_sales_data,
    build_star_schema,
    save_for_tableau,
    ensure_output_dir,
    contrast_ratio,
    make_high_dimensional_dataset,
)

sns.set_theme(style="whitegrid", context="talk")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
WEEK = "week-14"
OUTPUT_DIR = ensure_output_dir(WEEK)

clean, _ = clean_sales_data(introduce_quality_issues(make_base_sales(n=2600, seed=43), seed=43))
clean['month'] = clean['order_date'].dt.to_period('M').dt.to_timestamp()
```

### Actividad 1. QA minimo del dataset final
Verifica coherencia básica antes de presentar el dashboard.


```python
qa_checks = pd.DataFrame([
    {'check': 'No hay order_id nulos', 'passed': clean['order_id'].isna().sum() == 0},
    {'check': 'No hay ventas negativas', 'passed': (clean['sales'] >= 0).all()},
    {'check': 'Las fechas de envio no preceden a las de orden', 'passed': (clean['ship_date'] >= clean['order_date']).all()},
    {'check': 'Region completa', 'passed': clean['region'].isna().sum() == 0},
])
qa_checks
```

```python
rubric_template = pd.DataFrame([
    {'criterion': 'Calidad del dato', 'score_1_5': None, 'comment': ''},
    {'criterion': 'Validez de calculos', 'score_1_5': None, 'comment': ''},
    {'criterion': 'Claridad visual', 'score_1_5': None, 'comment': ''},
    {'criterion': 'Narrativa y recomendacion', 'score_1_5': None, 'comment': ''},
])
rubric_template
```

```python
defense_outline = pd.DataFrame([
    {'section': 'Problema', 'prompt': 'Que decision o pregunta justifica este dashboard?'},
    {'section': 'Datos', 'prompt': 'Que fuente usaste y con que limitaciones?'},
    {'section': 'Preparacion', 'prompt': 'Que limpieza o modelado fue indispensable?'},
    {'section': 'Analisis', 'prompt': 'Que hallazgos son fuertes y cuales exploratorios?'},
    {'section': 'Recomendacion', 'prompt': 'Que accion concreta se desprende de la evidencia?'},
])
defense_outline
```

```python
for name, df in {
    'qa_checks': qa_checks,
    'rubric_template': rubric_template,
    'defense_outline': defense_outline,
}.items():
    save_for_tableau(df, WEEK, name)
```

### Uso teorico de herramientas
- La tabla `qa_checks` traduce calidad técnica en criterios verificables.
- La `rubric_template` alinea la defensa con evaluación explícita.
- `defense_outline` convierte el proyecto final en una secuencia argumentativa clara.
