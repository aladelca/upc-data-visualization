```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = np.random.lognormal(mean=0, sigma=1, size=100000)



sns.histplot(data, bins=50)
```

```python
n = 100

muestras = []
for i in range(1000):
    muestra = np.random.choice(data, size=n, replace=True)
    muestras.append(muestra)
```

```python
sns.histplot([np.mean(muestra) for muestra in muestras], bins=50)
```
