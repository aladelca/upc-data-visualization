### Actividad 1. Entender la fuente antes de visualizar
En esta primera actividad no buscamos construir un gran dashboard, sino identificar:
- que representa una fila
- que variables son medidas y cuales son dimensiones
- que preguntas simples ya podemos responder


```python
summary = {
    'rows': len(sales),
    'columns': sales.shape[1],
    'date_min': sales['order_date'].min(),
    'date_max': sales['order_date'].max(),
    'regions': sales['region'].nunique(),
    'categories': sales['category'].nunique(),
}
pd.Series(summary)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
(
    sales.groupby('category', as_index=False)['sales']
    .sum()
    .sort_values('sales', ascending=False)
    .plot(kind='bar', x='category', y='sales', ax=axes[0], legend=False, color='#4b5563', title='Ventas por categoria')
)
monthly = sales.assign(month=sales['order_date'].dt.to_period('M').dt.to_timestamp())
monthly.groupby('month', as_index=False)['sales'].sum().plot(x='month', y='sales', ax=axes[1], color='#111827', title='Ventas mensuales')
axes[1].set_ylabel('sales')
plt.tight_layout()
```
