```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from _shared import (
    make_base_sales,
    introduce_quality_issues,
    profile_dataframe,
    clean_sales_data,
    build_star_schema,
    save_for_tableau,
    ensure_output_dir,
    contrast_ratio,
    make_high_dimensional_dataset,
)

sns.set_theme(style="whitegrid", context="talk")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
WEEK = "week-09"
OUTPUT_DIR = ensure_output_dir(WEEK)

clean, _ = clean_sales_data(introduce_quality_issues(make_base_sales(n=2200, seed=23), seed=23))
contrast_ratio('#111111', '#f3f4f6')
```

```python
contrast_examples = pd.DataFrame([
    {'foreground': '#111111', 'background': '#f3f4f6', 'ratio': contrast_ratio('#111111', '#f3f4f6')},
    {'foreground': '#6b7280', 'background': '#f3f4f6', 'ratio': contrast_ratio('#6b7280', '#f3f4f6')},
    {'foreground': '#9ca3af', 'background': '#ffffff', 'ratio': contrast_ratio('#9ca3af', '#ffffff')},
])
contrast_examples
```

```python
category_sales = clean.groupby('category', as_index=False)['sales'].sum().sort_values('sales', ascending=False)
fig, axes = plt.subplots(1, 2, figsize=(16, 6))
axes[0].bar(category_sales['category'], category_sales['sales'], color=['#d1d5db', '#9ca3af', '#6b7280'])
axes[0].set_title('Version debil: contraste y jerarquia limitados')
axes[1].bar(category_sales['category'], category_sales['sales'], color=['#111827', '#4b5563', '#9ca3af'])
axes[1].set_title('Version mejorada: foco y contraste')
plt.tight_layout()
```

```python
accessibility_checklist = pd.DataFrame([
    {'criterion': 'Contraste suficiente', 'status': 'Revisar'},
    {'criterion': 'No depender solo del color', 'status': 'Revisar'},
    {'criterion': 'Etiquetas legibles', 'status': 'Revisar'},
    {'criterion': 'Vista principal clara', 'status': 'Revisar'},
])
accessibility_checklist
```

```python
save_for_tableau(contrast_examples, WEEK, 'contrast_examples')
save_for_tableau(accessibility_checklist, WEEK, 'accessibility_checklist')
```
