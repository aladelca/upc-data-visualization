```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from _shared import (
    make_base_sales,
    introduce_quality_issues,
    profile_dataframe,
    clean_sales_data,
    build_star_schema,
    save_for_tableau,
    ensure_output_dir,
    contrast_ratio,
    make_high_dimensional_dataset,
)

sns.set_theme(style="whitegrid", context="talk")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
WEEK = "week-05"
OUTPUT_DIR = ensure_output_dir(WEEK)

clean, _ = clean_sales_data(introduce_quality_issues(make_base_sales(n=2200, seed=11), seed=11))
clean.head()
```

### Actividad 1. Comparacion, distribucion, relacion y tiempo
Usa el mismo dataset para responder cuatro preguntas distintas. La meta es justificar por qué cada gráfico es adecuado.


```python
fig, axes = plt.subplots(2, 2, figsize=(18, 12))
(
    clean.groupby('category', as_index=False)['sales'].sum()
    .sort_values('sales', ascending=False)
    .plot(kind='bar', x='category', y='sales', color='#111827', ax=axes[0,0], legend=False, title='Comparacion por categoria')
)
sns.histplot(clean['sales'], bins=30, color='#6b7280', ax=axes[0,1])
axes[0,1].set_title('Distribucion de ventas')
sample = clean.sample(600, random_state=11)
sns.scatterplot(data=sample, x='sales', y='profit', hue='category', palette='Greys', ax=axes[1,0])
axes[1,0].set_title('Relacion entre ventas y profit')
monthly = clean.assign(month=clean['order_date'].dt.to_period('M').dt.to_timestamp())
monthly.groupby('month', as_index=False)['sales'].sum().plot(x='month', y='sales', color='#374151', ax=axes[1,1], title='Serie temporal mensual')
plt.tight_layout()
```
